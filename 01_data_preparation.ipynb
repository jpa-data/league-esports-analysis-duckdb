## Loading and Exploring Data

```python
%sql SELECT * FROM read_csv('./data/raw/*.csv') LIMIT 3
```

We can see that the dataset contains many columns that are not needed for the analysis. 
The required columns for the team analysis are:
gameid, league, date, game, teamname, result
The columns for meta-trend analysis are:
gameid, league, date, playername, position, champion, ban1-5, gamelength
The additional columns for basic-player-stats analysis are:
kills, assists, deaths

```python
%%sql
CREATE OR REPLACE TABLE raw_league_data AS
SELECT gameid, league, date, patch, game, teamname, result, playername, position, champion, ban1, ban2, ban3, ban4, ban5, gamelength, kills, assists, deaths
FROM read_csv('./data/raw/*.csv')
```

```python
%%sql
SELECT * FROM raw_league_data LIMIT 3
```

Based on this information we can create tables for team and player data. The tables cast to the smallest appropriate datatypes and converts gamelength into minutes and seconds, which is much easier to read than stating the gamelength in total seconds. Bans have also been converted to a list instead of having 5 different columns to enable easier filtering. I have opted to keep result as an integer instead of turning it into a boolean as DuckDB does not support avg aggregation on booleans without casting it back into an integer each time.

```python
%%sql
CREATE OR REPLACE TABLE team_data  AS (
    WITH other_team AS(
    SELECT 
        gameid,
        teamname AS opponent
    FROM 
        raw_league_data
    WHERE 
        position = 'team'
    
)
    SELECT 
        rld.gameid, 
        date,
        patch,
        league,
        teamname,
        other_team.opponent,
        game::TINYINT as game,
        result::TINYINT as result,
        to_seconds(gamelength) as gamelength,
        list_value(ban1, ban2, ban3, ban4, ban5) as bans
    FROM 
        raw_league_data rld 
        JOIN other_team USING(gameid) 
    WHERE 
        rld.teamname != other_team.opponent 
        AND position = 'team'
)
```

Adding a split column and filling in accordingly. The dataset already contains a split column but due to inconsistent naming convention and different rules in some leagues I have created the column following the Winter/Spring/Summer convention and applied it to each region to have a consistent naming convention and thereby allow for accurate split aggregation. The end date of each split is the last game of the corresponding international event (First Stand for Winter, MSI for Spring, Worlds for Summer) and the start date of the following split is one day after or start of the year for winter (there are no games after the Worlds Tournament). 

```python
%%sql
ALTER TABLE team_data
ADD COLUMN split VARCHAR
```

```python
%%sql
UPDATE team_data
SET split = 
    CASE 
        WHEN strftime(date, '%Y-%m-%d') BETWEEN '2026-01-01' AND '2026-03-22' 
            OR strftime(date, '%Y-%m-%d') BETWEEN '2025-01-01' AND '2025-03-16' 
            THEN 'Winter'
        WHEN strftime(date, '%Y-%m-%d') BETWEEN '2026-03-23' AND '2026-07-12' 
            OR strftime(date, '%Y-%m-%d') BETWEEN '2025-03-17' AND '2025-07-12'  
            THEN 'Spring'
        ELSE 'Summer'
    END
```

```python
%%sql
SELECT * 
FROM team_data 
LIMIT 3
```

```python
%%sql 
CREATE OR REPLACE TABLE player_data AS (
    SELECT 
        gameid,
        teamname,
        playername,
        position,
        champion,
        kills::TINYINT as kills,
        assists::TINYINT as assists,
        deaths::TINYINT as deaths
    FROM 
        raw_league_data
    WHERE 
        position != 'team'
)
```

```python
%%sql
SELECT * FROM player_data LIMIT 3
```

We can join the tables based on the combination of gameid and teamname.

```python
%%sql
SELECT * FROM player_data JOIN team_data USING(gameid, teamname) LIMIT 3
```

## DATA CLEANING

For the analysis I am interested in the major leagues: LCK, LEC, LPL, LCS, CBLOL, LCP. The LCS and CBLOL underwent a rebranding in 2026 (LTA S -> CBLOL, LTA N -> LCS) and therefore have different identifiers in the 2025 dataset. The following maps the 2025 values to the currently active ones to prevent fractured historical data and inaccurate league aggregations. 

```python
%%sql
UPDATE 
    team_data
SET 
    league = CASE league 
        WHEN 'LTA S' THEN 'CBLOL'
        WHEN 'LTA N' THEN 'LCS'
        ELSE league
    END 
```

The data contains regional qualifying matches for Worlds 2025, these have the same league value as the main tournament, despite not being part of it. I have therefore changed the value of all games before the start of the main tournament to reflect that they are qualifying games and thereby removed them from aggregations specific to this tournament.

```python
%%sql 
SELECT * 
FROM 
    team_data 
WHERE 
    league = 'WLDs' 
    AND date BETWEEN '2025-01-01' AND '2025-10-13'
ORDER BY 
    date DESC
LIMIT 3
```

```python
%%sql
UPDATE 
    team_data
SET 
    league = 'qWLDs'
WHERE
    league = 'WLDs'
    AND date < '2025-10-14'
```

Starting in 2025, the naming convention for ingame patches has changed to use the year followed by patch number. This is not reflected in the dataset but can be easily fixed by adding 10.

```python
%%sql
SELECT DISTINCT patch FROM team_data ORDER BY patch LIMIT 3
```

```python
%%sql
UPDATE 
    team_data
SET
    patch = round(patch + 10, 2)
```

## Exporting Files
I have exported the results of these steps into parquet files and limited the leagues in both to those relevant for my analysis. The chosen leagues reflect the three official international events (MSI, First Stand, Worlds) and the 6 leagues that are able to qualify for them. This results in a drastically reduced file size as well as faster reading and querying speeds. The files are part of the repository and are in the data folder. Every analysis in 02_data_analysis is based on these files unless stated otherwise.

```python
%%sql
COPY (
    SELECT * 
    FROM team_data
    WHERE 
        league IN ('MSI', 'FST', 'WLDs', 'LPL', 'LCK', 'LCP', 'CBLOL', 'LCS', 'LEC') 
) TO './data/team_data.parquet' (FORMAT parquet, COMPRESSION zstd)
```

```python
%%sql
COPY (
    SELECT player_data.*
    FROM player_data JOIN team_data USING (gameid, teamname)
    WHERE 
        league IN ('MSI', 'FST', 'WLDs', 'LPL', 'LCK', 'LCP', 'CBLOL', 'LCS', 'LEC') 
) TO './data/player_data.parquet' (FORMAT parquet, COMPRESSION zstd)
```
